```python
import sys
from pathlib import Path
sys.path.append(str(Path.cwd().parent))

from src.data.loader import build_database, load_main_table
from src.data.preprocessor import clean, engineer_features, categorize_features
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_style('whitegrid')
```

```python
# Build the SQLite DB from the raw CSVs in ../data (run once)
build_database()
df = load_main_table()
df.shape
```

```python
# Data quality: missingness
missing = df.isna().mean().sort_values(ascending=False)
(missing.head(15) * 100).round(1)
```

```python
# Class balance — the key structural fact about this dataset
default_rate = df['TARGET'].mean() * 100
print(f'Default rate: {default_rate:.2f}%')
df['TARGET'].value_counts().plot(kind='bar', title='Class balance')
```

```python
# Business insight: default rate by education level
df.groupby('NAME_EDUCATION_TYPE')['TARGET'].mean().sort_values(ascending=False) * 100
```

```python
# Business insight: external bureau scores vs default
ext_cols = [c for c in df.columns if c.startswith('EXT_SOURCE')]
df[ext_cols + ['TARGET']].corr()['TARGET']
```

```python
# Feature categorization used later in the UI's EDA section
engineered = engineer_features(clean(df))
categorize_features(engineered)
```
